```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784')
X = mnist.data
y = mnist.target
```

```python
X
```

```python
df=pd.DataFrame(X)
df.insert(0,'label',y)
```

```python
df.head()
```

```python
df.insert(1,'pixel0',df['pixel9'])
```

```python
df
```

```python
df.head()
```

```python
df=df.iloc[:,:-1]
```

```python
df.head()
```

```python
X=df.iloc[:,1:]
y=df.iloc[:,0]
```

```python
sns.heatmap(X.iloc[5].values.reshape(28,28))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
sns.heatmap(rf.feature_importances_.reshape(28,28))
```

```python
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
```

```python
X,y=make_classification(n_samples=5,n_classes=2,n_features=2,n_informative=2,n_redundant=0,random_state=0)
```

```python
clf=DecisionTreeClassifier()
```

```python
clf.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf)
```

```python
clf.feature_importances_#1st col feature importance is more than that of second
```

```python
X,y=make_classification(n_samples=15,n_classes=2,n_features=2,n_informative=2,n_redundant=0,random_state=0)
```

```python
clf=DecisionTreeClassifier()
```

```python
clf.fit(X,y)
```

```python
plot_tree(clf)
```

```python
clf.feature_importances_
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=2)
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
print(rf.estimators_[0].feature_importances_)#feature importance according to 1st dt
print(rf.estimators_[1].feature_importances_)#feature importance according to 2st dt
```

```python
# 1+1/2=1
# 0+0/2=0
#here in random forest internally dt are used to get the feature importance of each columns
#only diffrence is in random forest average feature importance for each column is calculate using all dt used 
```
