```python
import numpy as np
import pandas as df
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import BaggingClassifier
```

```python
X,y=make_classification(n_samples=100,n_features=10,n_informative=3)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
dt=DecisionTreeClassifier(random_state=2)
dt.fit(X_train,y_train)
y_pred=dt.predict(X_test)
print("dt accuracy",accuracy_score(y_test,y_pred))
```

<h3>BAGGING</h3>

```python
bag=BaggingClassifier(
    estimator=DecisionTreeClassifier(),#base model
    n_estimators=500,#no of dt models
    max_samples=0.5,#no of rows for sampling
    bootstrap=True,#row sampling with replacement
    random_state=2
)
```

```python
bag.fit(X_train,y_train)
```

```python
y_pred=bag.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
bag.estimators_samples_[0].shape
```

```python
bag.estimators_features_[0].shape
```

<h3>BAGGING USING SVM</h3>

```python
bag=BaggingClassifier(
    estimator=SVC(),#base model
    n_estimators=500,#no of dt models
    max_samples=0.25,#no of rows for sampling
    bootstrap=True,#row sampling with replacement
    random_state=2
)
```

```python
bag.fit(X_train,y_train)
y_pred=bag.predict(X_test)
accuracy_score(y_test,y_pred)
```

<h3>PASTING

```python
bag=BaggingClassifier(
    estimator=DecisionTreeClassifier(),#base model
    n_estimators=500,#no of dt models
    max_samples=0.25,#no of rows for sampling
    bootstrap=False,#row sampling with replacement #only diffrence
    random_state=2,
    verbose=1,
    n_jobs=-1
)
```

```python
bag.fit(X_train,y_train)
y_pred=bag.predict(X_test)
accuracy_score(y_test,y_pred)
```

<h3>RANDOM SUBSPACES

```python
bag=BaggingClassifier(
    estimator=DecisionTreeClassifier(),#base model
    n_estimators=500,#no of dt models
    max_samples=1.0,#no of rows for sampling
   # bootstrap=False,#row sampling with replacement #only diffrence
    random_state=2,
    bootstrap_features=True,#colum sapling
    verbose=1,
    n_jobs=-1
)
```

```python
bag.fit(X_train,y_train)
y_pred=bag.predict(X_test)
accuracy_score(y_test,y_pred)
```

<h3>OOB SCORE</h3>

```python
bag=BaggingClassifier(
    estimator=DecisionTreeClassifier(),#base model
    n_estimators=500,#no of dt models
    max_samples=0.25,#no of rows for sampling
    bootstrap=True,#row sampling with replacement #only diffrence
    random_state=2,
    oob_score=True
   
)
```

```python
bag.fit(X_train,y_train)
```

```python
bag.oob_score_
```

```python
y_pred=bag.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```

```python
# Bagging Tips
# Bagging generally gives better results than Pasting
# Good results come around the 25% to 50% row sampling mark
# Random patches and subspaces should be used while dealing with high dimensional data
# To find the correct hyperparameter values we can do GridSearchCV/RandomSearchCV
```

```python
#appliying  GridSearchCV/RandomSearchCV
from sklearn.model_selection import GridSearchCV
```

```python
parameters={
    'n_estimators':[50,100,500],
    'max_samples':[0.1,0.4,0.7,1.0],
    'bootstrap':[True,False],
    'max_features':[0.1,0.4,0.7,0.1]
    
}
```

```python
search= GridSearchCV(BaggingClassifier(),parameters,cv=5)
```

```python
search.fit(X_train,y_train)
```

```python
search.best_params_
search.best_score_
```
